**BE CAREFUL COPYING ARRAYS**

```python
a = np.array([1,2,3])
b = a
b[0] = 100
print(a)
print(b)
#variable name b is now a so be careful
```

```python
a = np.array([1,2,3])
b = a.copy()
b[0] = 100
print(a)
print(b)
#using .copy() will only copy the values not the variable name
```
